```python
import os
from omegaconf import OmegaConf

import pandas as pd

from sklearn.metrics import accuracy_score
```

```python
# LABELS = ['1', '2', '3', '4', '5']
LABELS = ['1', '2', '3', '4']
```

```python
config = OmegaConf.load("../config.yaml")
```

```python
cls_result = [
    pd.read_csv(
        os.path.join("../../..", config.results_dir, f"cls_{target_label}_{pred_label}.csv")
    )
    for target_label in LABELS
    for pred_label in LABELS
]
cls_result = pd.concat(cls_result, ignore_index=True)
cls_result.head()
```

```python
pred_argmin = cls_result.groupby(["target", "traj_num"])["distance"].idxmin()
pred = cls_result.loc[pred_argmin]["pred"]
true = cls_result.loc[pred_argmin]["target"]
accuracy_score(true, pred)
```

```python
pred_df = cls_result.loc[pred_argmin]
for target, group_df in pred_df.groupby("target"):
    pred = group_df["pred"]
    true = group_df["target"]
    print(target, accuracy_score(true, pred))
```
